## Elo Rating System

### What is Elo?

- **Elo** is a **rating system** used to measure player strength in competitive games.
- Invented by **Arpad Elo**, a Hungarian–American physics professor.
- Commonly used in:
  - Chess (FIDE and online chess platforms)
  - Other board games (Go, shogi, etc., sometimes with tweaks)
  - Some online games and matchmaking systems

Each player has a **numerical rating**.  
Higher rating = stronger player (according to past game results).

---

### Core Idea

For a game between two players:

1. Use their ratings to compute the **expected score** of each player.
2. Compare the **actual result** to this **expected result**.
3. Update ratings:
   - If a player does **better than expected**, their rating goes **up**.
   - If they do **worse than expected**, their rating goes **down**.

Informally:

> If a weak player beats a strong player, it’s “surprising” → change ratings a lot.  
> If a strong player beats a weak player, it’s “expected” → change ratings a little.

---

### Formula (Conceptual)

Let:

- $ R_A $: rating of player A  
- $ R_B $: rating of player B  

**Expected score** of A:

$$
E_A = \frac{1}{1 + 10^{(R_B - R_A) / 400}}
$$

- If $R_A = R_B$, then $E_A \approx 0.5$ (equal strength).
- If $R_A \gg R_B$, then $E_A$ is close to 1 (A expected to win).
- If $R_A \ll R_B$, then $E_A$ is close to 0 (A expected to lose).

**Actual score** of A for one game:

- Win: $S_A = 1$  
- Draw: $S_A = 0.5$  
- Loss: $S_A = 0$

**Rating update** for A:

$$
R'_A = R_A + K (S_A - E_A)
$$

- $K$ is a constant (e.g., 10, 20, 32) that controls how fast ratings change.
- If $S_A > E_A$ → A did better than expected → rating increases.
- If $S_A < E_A$ → A did worse than expected → rating decreases.

Player B’s rating is updated similarly, usually in the opposite direction.

---

### Intuition Example

- Player A: 1600  
- Player B: 2000  

**If A wins**:

- Very surprising result.
- A’s rating goes up a lot.
- B’s rating goes down a lot.

**If B wins**:

- Expected result.
- Ratings change only slightly (small gain for B, small loss for A).
